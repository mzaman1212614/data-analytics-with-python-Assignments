# 4.Q

```python
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number] +=1
    else:
        frequency[number] = 1
        
print(frequency)       
```

# 5.Q

```python
sentence = input("Enter your sentence:")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] +=1
    else:
        frequency[word] =1
print(frequency)        
    
```
