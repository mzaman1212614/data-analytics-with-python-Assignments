3. Write a program in which use all the operators we can use in Python

```python
a = 10
b = 20
#  1.Arithematic Operators
print("------Arithematic Operators---")
print("Addition : " , a + b)
print("Subtraction : " , a - b)
print("Multiplication : ", a * b)
print("Division : ", a / b)
print("Floor Division : ", a // b)
print("Modulus : ", a % b)
print("Exponentiation : " , a ** b)

# 2.Comparison (Realtional) Operators
print("----Comparison Operators----")
print("Equal to : ", a==b)
print("Not Equal to : ", a!=b)
print("Greater Than : ", a > b)
print("Less Than : ", a < b)
print("Greater Than or Equal to : ", a>=b)
print("Less Than or Equal to : ", a<=b)

# 3.Logical Operators 
print("------Logical Operators-----")
x = True
y = False
print("AND :", x and y)
print("OR :" , x or y)
print("NOT :", not x)

# 4.Assignment Operators
print("------Assignment Operators----")
c = 2
print("Initial c :",c)
c += 3
print("c+= 3 :",c)
c -= 2
print("c-=2 :",c)
c *= 2
print("c*= :",c)
c /= 5
print("c/=5  :",c)
c //= 6
print("c//=6:",c)
c **= 2
print("c**=2 :",c)
c %= 5
print("c%=5 :",c)

# 5.Bitwise Operators
print("------Bitwise Operators-----")
p = 6 #110
q = 3 #011
print("Bitwise AND:", p & q)
print("Bitwise OR:", p | q)
print("Bitwise XOR:", p ^ q)
print("Bitwise NOT:", ~p)
print("Left Shift:", p << 1)
print("Right Shift:", p >> 1)

# 6.Identity Operators
print("------Identity Operators----")
m = [1,2,3]
n = [1,2,3]
o = m
print("m is n: " , m is n)
print("m is o:" , m is o)
print("m is not n:" ,m is not n)

# 7.Membership Operators
print("------Membership Operators----")
a = [1,3,4,5,6]
print("3 is in a :",3  in a)
print("8 is not in a:", 8  not in a)


```

4. Completes the following steps of small task:
    - Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
    - Mention Variable of Total Marks and assign 300 to it
    - Calculate Percentage

```python
english_marks = 56
islamiatMarks = 88
maths_marks = 45
total_marks = 300
obtained_marks = english_marks + islamiatMarks + maths_marks
percentage_of_marks = ( obtained_marks / total_marks ) * 100
print("English Marks is:" , english_marks)
print("Islamiat Marks is:", islamiatMarks)
print("Maths Marks is: " , maths_marks)
print("Obtained Marks is:", obtained_marks)
print("Percentage is:" ,percentage_of_marks,"%")
```

6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
# As we know that Area of circle = π * r^2 and circumference = 2 * π * r
r = int(input("Enter the radius of circle:"))
area_of_circle = 3.14159 * r * r
circumference_of_circle = 2 * 3.14159 * r
print("Area of circle is:", area_of_circle)
print("Circumference of circle is:", circumference_of_circle)


```

7. Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

```python
price = float(input("Enter price of an item : "))
discount_percentage = float(input("Enter discount percentage of an item : "))
discount_amount = (discount_percentage / 100) * price 
final_price = price - discount_amount 
print("Discount Amount : ", discount_amount )
print("Final Price : " , final_price )
```
