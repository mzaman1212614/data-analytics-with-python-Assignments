# 9.Q

```python
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

A = []
B = []

print("Enter Matrix A:")
for i in range(n):
    row = []
    for j in range(m):
        value = int(input(f"A[{i}][{j}]: "))
        row.append(value)
    A.append(row)

print("Enter Matrix B:")
for i in range(n):
    row = []
    for j in range(m):
        value = int(input(f"B[{i}][{j}]: "))
        row.append(value)
    B.append(row)

# Add matrices
C = []

for i in range(n):
    row = []
    for j in range(m):
        row.append(A[i][j] + B[i][j])
    C.append(row)

print("Result:")
for row in C:
    print(row)
```

# 10.Q

```python
n = int(input("Enter rows of Matrix A: "))
m = int(input("Enter columns of Matrix A: "))

p = int(input("Enter rows of Matrix B: "))
q = int(input("Enter columns of Matrix B: "))

if m != p:
    print("Matrix multiplication is not possible.")

else:
    A = []
    B = []

    print("Enter Matrix A:")
    for i in range(n):
        row = []
        for j in range(m):
            value = int(input(f"A[{i}][{j}]: "))
            row.append(value)
        A.append(row)

    print("Enter Matrix B:")
    for i in range(p):
        row = []
        for j in range(q):
            value = int(input(f"B[{i}][{j}]: "))
            row.append(value)
        B.append(row)

    # Create result matrix filled with 0
    C = []

    for i in range(n):
        row = []
        for j in range(q):
            row.append(0)
        C.append(row)

    # Matrix multiplication
    for i in range(n):
        for j in range(q):
            for k in range(m):
                C[i][j] = C[i][j] + A[i][k] * B[k][j]

    print("Result:")
    for row in C:
        print(row)
```
